### VggNet (layers: 16-19)

VGGNet (from the paper *“Very Deep Convolutional Networks for Large-Scale Image Recognition”*) is a convolutional neural network that became famous because:

✔️ It uses **very small filters (3×3)**

✔️ It stacks **many convolution layers** to get depth

✔️ It has a **simple, uniform architecture**

The most famous version is **VGG-16** (16 layers with weights).

---

🧱 1. Basic Building Block

**Conv(3×3) → ReLU**

- All convolution filters are **3×3**, stride = 1, padding = 1
- This keeps the **spatial size** (H×W) the same.
- Using two 3×3 convolutions instead of one 5×5 gives:
    - More nonlinearity
    - Fewer parameters
    - Better feature extraction

**MaxPool(2×2, stride 2)**

- Reduces height & width by **half** after each block.

---

🏗️ 2. VGGNet Architecture Layout

VGG-16 consists of **5 convolutional blocks**, followed by 3 fully connected layers.

---

🔵 **Block 1**

- Conv 3×3
- Conv 3×3
- MaxPool 2×2
    
    (Channels: 64 → 64)
    

---

🟢 **Block 2**

- Conv 3×3
- Conv 3×3
- MaxPool 2×2
    
    (Channels: 128 → 128)
    

---

🟡 **Block 3**

- Conv 3×3
- Conv 3×3
- Conv 3×3
- MaxPool 2×2
    
    (Channels: 256 → 256 → 256)
    

---

🟠 **Block 4**

- Conv 3×3
- Conv 3×3
- Conv 3×3
- MaxPool 2×2
    
    (Channels: 512 → 512 → 512)
    

---

🔴 **Block 5**

- Conv 3×3
- Conv 3×3
- Conv 3×3
- MaxPool 2×2
    
    (Channels: 512 → 512 → 512)
    

---

📦 After Convolution Blocks → Fully Connected Layers

Input image size = **224×224**

After 5 max pool layers (each halves size):

```
224 → 112 → 56 → 28 → 14 → 7

```

So before FC layers we have a **7×7×512** feature map.

---

🧩 **Fully Connected (FC)**

1. FC: 4096
2. FC: 4096
3. FC: 1000 (ImageNet classes)
4. Softmax

---

📚 Summary Table (VGG-16)

| Stage | Layers | Output size |
| --- | --- | --- |
| Input | – | 224×224×3 |
| Block 1 | 2×Conv(64) + Pool | 112×112×64 |
| Block 2 | 2×Conv(128) + Pool | 56×56×128 |

### Resnet (layers: 18, 34, 50, 101, 152)

**ResNet** is a deep CNN that uses **skip connections (residual connections)** to solve the vanishing gradient problem and allow networks to become very deep (50–152 layers).

**Structure**

- Initial: 7×7 conv → maxpool
- 4 main stages (Conv2_x → Conv5_x)
- Each stage contains multiple **residual blocks**
- Two block types:
    - **Basic block** (ResNet-18/34): 3×3 → 3×3 + skip
    - **Bottleneck block** (ResNet-50/101/152): 1×1 → 3×3 → 1×1 + skip

**Skip Connection**

- Input **bypasses** the conv layers and is **added** to the block output:
    
    `Output = F(x) + x`
    
- Helps gradients flow backward → easier training → deeper networks.

**Bottleneck**

- 1×1 reduce → 3×3 conv → 1×1 expand
- Makes deep models **faster and lighter**, not slower.
- Greatly reduces compute while keeping accuracy high.

**Why ResNet > AlexNet/VGG**

- Much deeper but easier to train
- Better accuracy
- More efficient (fewer parameters than VGG)
- Extremely good for transfer learning.

**When to use**

- For almost all image tasks: classification, detection, segmentation, medical images, etc.

### DenseNet (layers: 121, 169, 201, 264)

**Core idea:**

Every layer receives **all previous layers’ feature maps** as input and **concatenates** them, instead of just taking the previous layer’s output. This dense connectivity improves gradient flow and feature reuse.

**Why Dense Connections?**

- **Better gradient flow:** eases training of deep networks
- **Feature reuse:** earlier features are reused, reducing redundancy
- **Fewer parameters:** small growth rate (12–32) keeps the model efficient
- **High accuracy:** compact yet powerful

**Structure**

- **Dense Blocks:** layers connected to all previous layers via concatenation
- **Transition Layers:** 1×1 conv + 2×2 avg pooling to reduce channels and spatial size

**Pros:**

- Excellent gradient flow
- High accuracy with fewer parameters
- Feature reuse → less overfitting

**Cons:**

- High memory usage (concatenation)
- Slower inference
- More complex implementation

**When to Use**

- Best for: high accuracy, feature reuse, medical imaging, small datasets
- Avoid when: fast inference needed, limited GPU memory, very large images

### EfficientNet

- **Core Idea (Compound Scaling)**: The fundamental idea behind EfficientNet is **compound scaling**. Instead of scaling network depth, width, or resolution independently, EfficientNet scales all three dimensions simultaneously and uniformly using a fixed set of scaling coefficients. This allows for a more optimal balance and leads to significantly better performance and efficiency compared to conventional scaling methods.
- **Dense Connections**: It's important to clarify that **EfficientNet does not directly employ the 'dense connections'** found in architectures like DenseNet, where each layer is connected to every subsequent layer in a feed-forward fashion. Instead, EfficientNet uses efficient building blocks called **MBConv (Mobile Inverted Bottleneck Convolution) blocks**. These blocks are highly efficient and incorporate:
    - **Depthwise separable convolutions** to reduce computational cost.
    - **Squeeze-and-Excitation (SE) blocks** which adaptively recalibrate channel-wise feature responses. While SE blocks introduce a form of attention mechanism that enhances feature utility, they are not 'dense connections' in the traditional sense of connecting layers directly to all previous layers.
- **Structure**: EfficientNet's architecture is built upon a baseline network called **EfficientNet-B0**. All other EfficientNet models (B1 to B7) are scaled versions of B0. The primary building block is the **MBConv block**, which features an inverted residual structure. The models also use **Swish activation functions** for improved performance.
- **Pros**:
    - **High Efficiency and Accuracy**: Achieves state-of-the-art accuracy with significantly fewer parameters and FLOPs than previous models.
    - **Systematic Scaling**: The compound scaling method provides a principled way to scale models, making them more robust and performant across different resource constraints.
    - **Transfer Learning**: Pre-trained EfficientNet models are excellent feature extractors and perform very well in transfer learning tasks.
- **Cons**:
    - **Complexity of Scaling**: While systematic, understanding and implementing the compound scaling from scratch can be complex.
    - **Memory Usage**: For larger versions (e.g., B7), memory requirements can still be substantial, especially for high-resolution inputs.
    - **Training Time**: Although efficient in terms of parameters, training from scratch can still be time-consuming for the larger models.
- **Number of Layer Versions**: EfficientNet has **8 main versions (models)**, denoted as **EfficientNet-B0 through EfficientNet-B7**. Each successive version represents a larger model with increased depth, width, and input resolution, scaled according to the compound scaling rule, offering a trade-off between accuracy and computational cost.
