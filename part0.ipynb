### What is Compose is used for?

`transforms.Compose`는 여러 개의 데이터 변환(transformation) 작업을 **순서대로 묶어 하나의 파이프라인으로 정의**하는 데 사용됩니다

### Where does Normalize values come from in augmentation?

1. **첫 번째 리스트** (`[0.485, 0.456, 0.406]`)는 정규화에 사용되는 **평균값(Mean)**입니다.

2. **두 번째 리스트** (`[0.229, 0.224, 0.225]`)는 정규화에 사용되는 **표준 편차(Standard Deviation)**입니다

### What is the difference between loading image using Image and cv2?

🧾 Summary Table

| Feature | PIL.Image.open() | cv2.imread() |
| --- | --- | --- |
| Color order | RGB | BGR |
| Return type | `PIL.Image.Image` | `numpy.ndarray` |
| Lazy load | Yes | No |
| Can read from bytes/stream | Yes | No |
| Primary use | Image editing/display | Image analysis/CV |

🧪 Practical Tip

If you’re working in **deep learning or computer vision** (e.g., PyTorch, OpenCV, TensorFlow):

- Prefer **OpenCV (cv2)** or convert to NumPy early.

If you’re working in **image manipulation, resizing, or format conversion**:

- Prefer **PIL (Pillow)**.
