### What is Compose is used for?

`transforms.Compose`는 여러 개의 데이터 변환(transformation) 작업을 **순서대로 묶어 하나의 파이프라인으로 정의**하는 데 사용됩니다

### Where does Normalize values come from in augmentation?

1. **첫 번째 리스트** (`[0.485, 0.456, 0.406]`)는 정규화에 사용되는 **평균값(Mean)**입니다.

2. **두 번째 리스트** (`[0.229, 0.224, 0.225]`)는 정규화에 사용되는 **표준 편차(Standard Deviation)**입니다
