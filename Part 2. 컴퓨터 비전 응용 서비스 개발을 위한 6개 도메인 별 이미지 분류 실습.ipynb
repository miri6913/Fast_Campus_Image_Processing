### What is the difference between over, undersampling and augmentation?

| Method | What It Does | Creates New Samples? | Pros | Cons |
| --- | --- | --- | --- | --- |
| **Oversampling** | Duplicate minority samples | ❌ (just copies) | Simple | Overfitting risk |
| **Undersampling** | Remove majority samples | ❌ | Fast | Lose data |
| **Augmentation** | Transform existing samples | ✅ | More variety | Needs careful design |

### SimCLR

**SimCLR (Simple Framework for Contrastive Learning of Visual Representations)** is a **self-supervised learning** method for learning image features **without labels**.

**Goal**: Learn useful visual representations by bringing *similar views of the same image* closer and pushing *different images* apart in embedding space.

**Core idea (contrastive learning)**

- Take one image
- Apply **two different random augmentations** (crop, color jitter, blur, etc.)
- Treat them as a **positive pair**
- All other images in the batch are **negative pairs**

**Architecture**

- **Encoder**: CNN (e.g., ResNet) → feature vector
- **Projection head**: small MLP used *only during training*
- Final representations are taken **before** the projection head

**Loss function**

- **NT-Xent (Normalized Temperature-scaled Cross Entropy Loss)**
- Encourages similarity for positive pairs, dissimilarity for negatives

**Strong data augmentation is critical**

- Performance depends heavily on augmentation choice and strength

**Large batch size**

- Needed to get many negative samples
- Alternatively uses tricks like memory banks in later variants

**After training**

- Encoder is reused for downstream tasks (classification, detection, etc.)
- Fine-tuning or linear probing is applied

**Why it matters**

- Achieves performance close to supervised learning
- Simple, effective baseline for self-supervised vision learning

### Self-Supervised learning

learning paradigm where **labels are generated from the data itself**, so no manual annotation is required.

### Latent space

A latent space, also known as a latent feature space or embedding space, is **an embedding of a set of items within a manifold in which items resembling each other are positioned closer to one another**.

### Loss function in ssl

SimCLR uses **NT-Xent contrastive loss** to maximize similarity between augmented views of the same image while minimizing similarity to all other images in the batch.

### Why do we use resnet18 or 50 in simclr?

They provide a good balance between representation quality, training stability, and computational cost for contrastive learning.

### NT-Xent (Normalized Temperature - Cross-entropy loss) detailed explanation

NT-Xent turns contrastive learning into a classification problem where the model must identify the correct augmented twin among many distractors.https://chatgpt.com/share/695b73e6-1f58-8005-9e73-570e5bf9142b

### Cosine annealing

Deep learning learning rate (LR) scheduling technique that smoothly decreases the LR from a maximum to a minimum using a half-cosine curve over a set period, helping models converge better by allowing large steps initially and fine-tuning later, often combined with warm restarts to escape local minima and improve generalization.

### Project embeddings, embedding networks

Projected embeddings are feature vectors produced by passing encoder outputs through a projection head, creating a space specifically used for the training objective (e.g., contrastive loss), not for downstream tasks.An embedding network is a neural network that maps input data into a fixed-dimensional vector space where similar inputs are close together and dissimilar inputs are far apart.

### Encoder

The neural network learning the features (e.g., ResNet).An encoder is the part of a model that takes raw input (e.g., an image) and converts it into a compact feature representation that captures its essential information.

### What is the difference between get_data_loaders and get_train_validation_data_loaders?

get_data_loaders() prepares the dataset and calls the process, while get_train_validation_data_loaders() does the actual work of splitting the data and building the PyTorch DataLoaders.
