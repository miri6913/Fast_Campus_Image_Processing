```python
import os
import random
import shutil
```

```python
data_root = 'C:/Users/Jarvis/Downloads/kaggle_fer_dataset-20251130T045347Z-1-001/kaggle_fer_dataset'

file_root = f'{data_root}/train1'
train_root = f'{data_root}/train'
valid_root = f'{data_root}/valid'
test_root = f'{data_root}/test'


for r in [train_root, valid_root, test_root]:
    if not os.path.exists(r):
        os.makedirs(r)

cls_list = os.listdir(file_root)

for cls in cls_list:
    train_cls_path = f"{train_root}/{cls}"
    if not os.path.exists(train_cls_path):
        os.makedirs(train_cls_path)
    valid_cls_path = f"{valid_root}/{cls}"
    if not os.path.exists(valid_cls_path):
        os.makedirs(valid_cls_path)
    
```

```python
from tqdm import tqdm
import random
import shutil
import os

random.seed(2024)

# Count total files for progress bar
total_files = sum(len(os.listdir(f"{file_root}/{cls}")) for cls in cls_list)

with tqdm(total=total_files, desc="Total file copy progress") as pbar:
    for cls in cls_list:
        file_list = os.listdir(f"{file_root}/{cls}")
        random.shuffle(file_list)

        test_ratio = 0.1
        valid_list = file_list[:int(len(file_list) * test_ratio)]
        train_list = file_list[int(len(file_list) * test_ratio):]

        for i in valid_list:
            shutil.copyfile(f"{file_root}/{cls}/{i}", f"{valid_root}/{cls}/{i}")
            pbar.update(1)

        for i in train_list:
            shutil.copyfile(f"{file_root}/{cls}/{i}", f"{train_root}/{cls}/{i}")
            pbar.update(1)
```
