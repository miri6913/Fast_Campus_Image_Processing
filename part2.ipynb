### What is the difference between over, undersampling and augmentation?

| Method | What It Does | Creates New Samples? | Pros | Cons |
| --- | --- | --- | --- | --- |
| **Oversampling** | Duplicate minority samples | ❌ (just copies) | Simple | Overfitting risk |
| **Undersampling** | Remove majority samples | ❌ | Fast | Lose data |
| **Augmentation** | Transform existing samples | ✅ | More variety | Needs careful design |
